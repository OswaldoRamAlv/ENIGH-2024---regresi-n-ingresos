Dividimos por viviendas para evitar el data leakage en el test set

```python
from sklearn.model_selection import GroupShuffleSplit

splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=55)

X = df.drop(columns=[
    "ing_tri_total", "folioviv",
    "foliohog", "numren", "id_per"
])

y = df["ing_tri_total"]
```

```python
train_idx, test_idx = next(splitter.split(X, y, groups=df["folioviv"]))

X_train = X.iloc[train_idx]
y_train = y.iloc[train_idx]
X_test = X.iloc[test_idx]
y_test = y.iloc[test_idx]
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
 
num_pipeline = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler()
)

preprocessor = ColumnTransformer([
    ("num", num_pipeline, ['edad', 'htrab_total', 'nivelaprob']),
], remainder="passthrough")


lin_reg_pipeline = make_pipeline(
    preprocessor,
    LinearRegression())
```

```python
lin_reg_pipeline.fit(X_train, y_train)
```

```python
ing_tri_pred = lin_reg_pipeline.predict(X_train)
ing_tri_pred[:5].round(2)
```

```python
y_train[:5].round(2)
```

### Aplicamos Metricas

en nuestro caso principalmente nos queremos enfocar en MAE ya que nos puede dar un error mas interpretable y mas representativo
ya que $r^2$ puede verse muy penalizado por lo valores atipicos

```python
from sklearn.metrics import root_mean_squared_error, r2_score, mean_absolute_error

lin_rmse = root_mean_squared_error(y_train, ing_tri_pred)
lin_r2 = r2_score(y_train, ing_tri_pred)
lin_mae = mean_absolute_error(y_train, ing_tri_pred)

print(f"RMSE: {lin_rmse:.2f}")
print(f"R2: {lin_r2:.2f}")
print(f"MAE: {lin_mae:.2f}")
```

Comprobamos en el test para saber si no hay un sobreajuste

```python
ing_tri_pred = lin_reg_pipeline.predict(X_test)

lin_rmse = root_mean_squared_error(y_test, ing_tri_pred)
lin_r2 = r2_score(y_test, ing_tri_pred)
lin_mae = mean_absolute_error(y_test, ing_tri_pred)

print(f"RMSE: {lin_rmse:.2f}")
print(f"R2: {lin_r2:.4f}")
print(f"MAE: {lin_mae:.2f}")
```

## Puntos a trabajar
El 23% de ceros necesita un tratamiento propio. Lo más limpio es un modelo en dos etapas: primero clasificar si tiene ingreso, y luego regresar el monto solo entre quienes sí tienen.

```python
from sklearn.tree import DecisionTreeRegressor

tree_reg = make_pipeline(preprocessor, DecisionTreeRegressor(random_state=42))
tree_reg.fit(X_train, y_train)
```

```python
tree_pred = tree_reg.predict(X_train)

tree_mae = mean_absolute_error(y_train, tree_pred)

print(f"Decision Tree Train MAE: {tree_mae:.2f}")
```

```python
tree_pred_test = tree_reg.predict(X_test)

tree_mae_test = mean_absolute_error(y_test, tree_pred_test)

print(f"Decision Tree Test MAE: {tree_mae_test:.2f}")
```
