## Unificar bases 

```python
#Crear identificador
poblacion['id_per'] = poblacion['folioviv'] + poblacion['foliohog'] + poblacion['numren'] 
ingreso['id_per']= ingreso['folioviv'] + ingreso['foliohog'] + ingreso['numren']
trabajos['id_per'] = trabajos['folioviv'] + trabajos['foliohog'] + trabajos['numren']


print('======== Poblacion ==========')
print(poblacion.head(3))
print('============== ingresos ==========')
print(ingreso.head(3))
print('============ Trabajos ===========')
print(trabajos.head(3))
```

```python
# Verificar si 'id_per' es único en cada tabla
print("¿Único en personas?:", poblacion["id_per"].is_unique)
print("¿Único en ingresos?:", ingreso["id_per"].is_unique)
print("¿Único en trabajos?:", trabajos["id_per"].is_unique)
```

#### Nota sobre ingresos y trabajos

Se pueden dar los casos en los que una persona pueda tener mas de un ingreso y mas de un trabajo.

(relacion de uno a muchos)

Por lo que se buscara la forma de cuantificar cuantos trabajos y cuantos ingresos tiene, para agrtegarlos como variable independiente, ademas de que se buscara sumarlos para que en la base final solo se tenga una entrada por persona

```python
# 1. Agrupar y agregar la tabla de ingresos
ingreso_agrupado = ingreso.groupby('id_per').agg(
    ing_tri_total=('ing_tri', 'sum'),      # Suma de todos los ingresos
    num_ingresos=('clave', 'count')        # Conteo de fuentes de ingreso
).reset_index()

# 2. Agrupar y agregar la tabla de trabajos
trabajos_agrupado = trabajos.groupby('id_per').agg(
    htrab_total=('htrab', 'sum'),          # Suma de horas trabajadas totales
    num_trabajos=('id_trabajo', 'count')   # Conteo de trabajos
).reset_index()

# 3. Unir las bases (Merge left para conservar a toda la poblacion)
df_final = poblacion.merge(ingreso_agrupado, on='id_per', how='left')
df_final = df_final.merge(trabajos_agrupado, on='id_per', how='left')

# 4. Rellenar nulos con 0 para personas sin trabajo/ingreso
cols_a_rellenar = ['ing_tri_total', 'num_ingresos', 'htrab_total', 'num_trabajos']
df_final[cols_a_rellenar] = df_final[cols_a_rellenar].fillna(0)

# Verificar unicidad y tamaño
print("=== Tamaño base unificada ===")
print(df_final.shape)
print("¿Único en base unificada?:", df_final['id_per'].is_unique)
df_final.head()
```

```python
# Convertir discapacidad en valores 0 y 1 (False True)
cols_disc = ['disc_ver', 'disc_oir', 'disc_brazo', 'disc_camin', 
             'disc_apren', 'disc_vest', 'disc_habla', 'disc_acti']
for col in cols_disc:
    df_final[col] = df_final[col].apply(lambda x: 0 if str(x).strip() == '1' else 1)

df_final.head()
```

```python
# Variables dummie para sexo

df_final['sexo'] = df_final['sexo'].replace({'1': 'H', '2': 'F'})

# Generar las variables dummy
df_final = pd.get_dummies(df_final, columns=['sexo'], drop_first=True, dtype=int)

df_final
```

```python
# Eliminar personas menores de 12 años
df_final = df_final[df_final['edad'] >= 12].copy()

# Verificar el cambio en el número de filas
print("=== Tamaño final ===")
print(df_final.shape)

df_final.head()
```

# exploracion inicial y eliminacion de atipicos


```python
# Histograma de ingresos antes de eliminacion de atipicos
sns.histplot(df_final['ing_tri_total'], kde=True)
```

```python
# IQR para eliminar outliers

Q1 = df_final['ing_tri_total'].quantile(0.25)
Q3 = df_final['ing_tri_total'].quantile(0.75)
IQR = Q3 - Q1
df_final = df_final[(df_final['ing_tri_total'] >= Q1 - 1.5 * IQR) & (df_final['ing_tri_total'] <= Q3 + 1.5 * IQR)]
```
